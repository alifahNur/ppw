# Crowling Web
**Web Crawling** (atau *web spidering*) adalah proses otomatisasi untuk mengumpulkan data dari situs web. Proses ini dilakukan oleh perangkat lunak yang dikenal sebagai *web crawler*, *spider*, atau *bot*. Berikut adalah penjelasan mendetail tentang web crawling:

### 1. **Apa itu Web Crawling?**
Web crawling adalah teknik yang digunakan untuk menjelajahi dan mengindeks konten yang ada di internet. Crawler mengunjungi halaman web, membaca konten, dan mengekstrak informasi yang relevan sesuai dengan tujuan pengumpulan data mereka.

### 2. **Bagaimana Cara Kerja Web Crawling?**
- **Mulai dari URL Awal**: Crawler biasanya dimulai dari satu atau beberapa URL awal (dikenal sebagai *seed URLs*).
- **Mengambil Konten Halaman**: Crawler mengirimkan permintaan HTTP ke URL tersebut dan mengunduh HTML dari halaman.
- **Mengekstrak Tautan**: Setelah halaman diunduh, crawler menganalisis konten untuk menemukan tautan (link) ke halaman lain.
- **Mengunjungi Halaman Tautan**: Tautan yang ditemukan kemudian ditambahkan ke dalam daftar URL yang harus dikunjungi. Proses ini berulang, sehingga crawler dapat menjelajahi seluruh web.
- **Penyimpanan dan Pengindeksan**: Informasi yang diambil dari halaman web disimpan dalam basis data atau indeks untuk analisis lebih lanjut atau pencarian.

### 3. **Penggunaan Web Crawling**
Web crawling memiliki berbagai aplikasi, antara lain:
- **Pencarian Informasi**: Mesin pencari seperti Google menggunakan crawler untuk mengindeks halaman web sehingga mereka dapat ditampilkan dalam hasil pencarian.
- **Pengumpulan Data**: Peneliti dan pengembang menggunakan crawling untuk mengumpulkan data dari situs web untuk analisis, riset pasar, dan pengembangan produk.
- **Monitoring**: Beberapa perusahaan menggunakan crawler untuk memantau perubahan di situs web kompetitor, termasuk harga dan penawaran produk.
- **Analisis Konten**: Crawler dapat digunakan untuk menganalisis konten dan sentimen di berbagai platform media sosial atau berita.

### 4. **Etika dan Kebijakan dalam Web Crawling**
- **Robots.txt**: Banyak situs web menggunakan file `robots.txt` untuk memberi tahu crawler tentang bagian mana dari situs yang boleh atau tidak boleh diakses. Crawler yang baik akan mematuhi aturan ini.
- **Batasan Lalu Lintas**: Untuk mencegah overloading server, crawler biasanya menerapkan batasan kecepatan pada seberapa cepat mereka mengunjungi halaman.
- **Privasi dan Data**: Pengumpulan data harus dilakukan dengan memperhatikan privasi pengguna dan hukum yang berlaku, terutama ketika mengumpulkan data sensitif.

### 5. **Teknologi yang Digunakan**
Web crawling biasanya dilakukan menggunakan bahasa pemrograman seperti Python, Java, atau Ruby. Beberapa pustaka yang umum digunakan dalam web crawling antara lain:
- **BeautifulSoup**: Untuk parsing HTML dan XML.
- **Scrapy**: Kerangka kerja yang dirancang khusus untuk web scraping dan crawling.

```python
#link dari berita yang diambil
url ='https://www.liputan6.com/news'
response = requests.get(url)
```

- requests.get(url): Fungsi get() dari pustaka requests digunakan untuk mengirimkan permintaan HTTP dengan metode GET ke URL yang telah didefinisikan sebelumnya. Metode GET digunakan untuk mengambil data dari server.
- Hasil dari permintaan ini disimpan dalam variabel response.
- Setelah mengirimkan permintaan, objek response yang dikembalikan akan berisi berbagai informasi tentang hasil permintaan, termasuk:


```python
#mengubah konten html menjadi objek
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
#menampilkan judul berita

# Temukan elemen <div> dengan kelas 'overflow margin-bottom-xs'
judul = soup.find('div', class_='articles')

# Jika elemen ditemukan, cari semua elemen <span> di dalamnya dan tampilkan teksnya
if judul:
    # Inisialisasi counter
    counter = 1

    for span in judul.find_all('span', class_= "articles--iridescent-list--text-item__title-link-text"):
        print(f"{counter}. {span.text}")
        counter += 1
else:
    print("Elemen dengan kelas 'overflow margin-bottom-xs' tidak ditemukan.")
```

### 1. **Mencari Elemen `<div>`**

- **`soup.find()`**: Fungsi ini digunakan untuk menemukan elemen pertama dalam dokumen HTML yang cocok dengan kriteria yang ditentukan.
- **`'div'`**: Menunjukkan bahwa kita mencari elemen `<div>`.
- **`class_='articles'`**: Menyaring pencarian untuk hanya elemen `<div>` yang memiliki atribut kelas `articles`.
- **`judul`**: Variabel ini akan menyimpan elemen `<div>` yang ditemukan. Jika elemen tidak ditemukan, `judul` akan bernilai `None`.

### 2. **Memeriksa dan Mencari Elemen `<span>`**

- **`if judul:`**: Memeriksa apakah elemen `<div>` dengan kelas `articles` ditemukan. Jika ya, maka blok kode berikutnya akan dijalankan.
- **`counter = 1`**: Inisialisasi sebuah variabel `counter` untuk menghitung jumlah elemen `<span>` yang ditemukan.
- **`judul.find_all('span', class_="articles--iridescent-list--text-item__title-link-text")`**:
  - **`find_all()`**: Fungsi ini digunakan untuk menemukan semua elemen `<span>` di dalam elemen `<div>` yang sudah ditemukan (yaitu `judul`).
  - **`class_="articles--iridescent-list--text-item__title-link-text"`**: Menyaring pencarian untuk hanya elemen `<span>` yang memiliki atribut kelas tertentu.
- **`for span in ...`**: Mengiterasi setiap elemen `<span>` yang ditemukan.
- **`print(f"{counter}. {span.text}")`**: Mencetak teks dari setiap elemen `<span>`, di depan teks tersebut ditambahkan nomor urut dari `counter`.
- **`counter += 1`**: Mengincrement nilai `counter` untuk setiap iterasi, sehingga nomor urut bertambah.

### 3. **Pesan Jika Tidak Ditemukan**

- Jika elemen `<div>` dengan kelas `articles` tidak ditemukan, program akan mencetak pesan yang menyatakan bahwa elemen tersebut tidak ditemukan.

### Kesimpulan
Kode ini bertujuan untuk mengekstrak dan mencetak teks dari elemen `<span>` yang terletak di dalam elemen `<div>` tertentu. Jika elemen `<div>` tersebut tidak ada, maka kode ini akan memberikan umpan balik kepada pengguna bahwa elemen yang dicari tidak ditemukan.

```python
#mengektrak dan mencetak link halaman berita
link_counter = 1
print("\nDaftar Link:")
seen_links = set()  # Set untuk melacak link yang sudah dicetak
for link in judul.find_all('a',class_="ui--a articles--iridescent-list--text-item__title-link"):
  href = link.get('href')
  if href and href not in seen_links:  # Periksa apakah href ada dan belum dicetak
    print(f"{link_counter}. {href}")
    seen_links.add(href)  # Tambahkan href ke set
    link_counter += 1
```


### 1. **Inisialisasi Variabel**

- **`link_counter = 1`**: Inisialisasi variabel `link_counter` untuk menghitung jumlah tautan yang ditemukan dan akan dicetak. Dimulai dari 1.
- **`print("\nDaftar Link:")`**: Mencetak judul untuk daftar tautan.
- **`seen_links = set()`**: Membuat set kosong bernama `seen_links` untuk menyimpan tautan yang telah dicetak. Penggunaan set menghindari duplikasi tautan.

### 2. **Mencari Elemen `<a>` dan Mencetak Tautan**

- **`for link in judul.find_all('a', class_="ui--a articles--iridescent-list--text-item__title-link"):`**
  - **`find_all()`**: Fungsi ini mencari semua elemen `<a>` (tautan) di dalam elemen `<div>` yang disimpan dalam variabel `judul`.
  - **`class_="ui--a articles--iridescent-list--text-item__title-link"`**: Mencari elemen `<a>` yang memiliki kelas tertentu.
  
- **`href = link.get('href')`**:
  - Mengambil nilai atribut `href` dari elemen `<a>`, yang merupakan URL tautan.

- **`if href and href not in seen_links:`**
  - Memeriksa apakah `href` tidak kosong dan belum ada di set `seen_links`. Ini menghindari pencetakan tautan yang sama lebih dari sekali.
  
- **`print(f"{link_counter}. {href}")`**:
  - Mencetak tautan yang ditemukan dengan nomor urut yang diambil dari `link_counter`.

- **`seen_links.add(href)`**:
  - Menambahkan tautan `href` ke set `seen_links` untuk melacak tautan yang sudah dicetak.

- **`link_counter += 1`**:
  - Mengincrement nilai `link_counter` untuk setiap tautan yang berhasil dicetak, sehingga nomor urut bertambah untuk tautan berikutnya.

### Kesimpulan
Kode ini digunakan untuk mengekstrak dan mencetak daftar tautan dari elemen yang memiliki kelas tertentu dalam dokumen HTML. Dengan menggunakan set untuk melacak tautan yang sudah dicetak, kode ini memastikan tidak ada tautan duplikat yang ditampilkan dalam daftar. Setiap tautan dicetak dengan nomor urut yang unik.

```python
#mengumpulkan dan menyimpan data artikel dari berita dalam format dataframe
import pandas as pd
from IPython.display import display  # Jika Anda menggunakan Jupyter Notebook

# List untuk menyimpan data artikel
data_artikel = []

# Mengambil semua artikel dari halaman
articles = soup.find_all('article')  # Ubah selector menjadi 'article' jika setiap artikel dibungkus dalam elemen <article>

# Set untuk menyimpan href yang sudah dilihat agar tidak terjadi duplikasi
seen_links = set()

# Iterasi melalui setiap artikel
for article in articles:
    # Ambil link artikel
    link = article.find('a', class_='ui--a articles--iridescent-list--text-item__title-link')
    if link:
        href = link.get('href')
        title = link.get('title').strip()

        if href and href not in seen_links:
            # Ambil tanggal publikasi
            date = article.find('time', class_='articles--iridescent-list--text-item__time')
            date_text = date.text.strip() if date else 'N/A'

            # Ambil kategori artikel
            category = article.find('a', class_='articles--iridescent-list--text-item__category')
            category_text = category.text.strip() if category else 'N/A'

            # Ambil ringkasan artikel
            summary_div = article.find('div', class_='articles--iridescent-list--text-item__summary articles--iridescent-list--text-item__summary-seo')
            summary_text = summary_div.text.strip() if summary_div else 'N/A'

            # Menyimpan data ke dalam list sebagai satu baris per artikel
            data_artikel.append([title, date_text, summary_text, category_text])

            # Tambahkan href ke set untuk menghindari duplikasi
            seen_links.add(href)

# Membuat DataFrame dari list data_artikel dengan kolom yang diinginkan
df = pd.DataFrame(data_artikel, columns=['Judul', 'Tanggal', 'Ringkasan', 'Kategori'])

# Menampilkan DataFrame menggunakan display
display(df)
```

- **`import pandas as pd`**: Mengimpor library Pandas yang digunakan untuk manipulasi dan analisis data, terutama untuk menyimpan data dalam format tabel.
- **`from IPython.display import display`**: Mengimpor fungsi `display` yang digunakan untuk menampilkan objek dengan baik di Jupyter Notebook.

### 2. **Inisialisasi List untuk Menyimpan Data**

- **`data_artikel = []`**: Membuat list kosong untuk menyimpan data artikel yang akan diambil dari halaman web.

### 3. **Mengambil Semua Artikel dari Halaman**

- **`soup.find_all('article')`**: Mengambil semua elemen `<article>` dari objek BeautifulSoup yang berisi HTML dari halaman web. Ini mengasumsikan bahwa setiap artikel dibungkus dalam elemen `<article>`. Jika tidak, Anda harus menggunakan selector yang sesuai.

### 4. **Set untuk Menyimpan Link yang Sudah Dilihat**

- **`seen_links = set()`**: Membuat set kosong untuk melacak tautan yang telah diproses, sehingga menghindari duplikasi saat mengambil data artikel.

### 5. **Iterasi Melalui Setiap Artikel**

- Memulai loop untuk iterasi melalui setiap elemen artikel yang ditemukan sebelumnya.

### 6. **Mengambil Data dari Setiap Artikel**

- **`article.find('a', class_='ui--a articles--iridescent-list--text-item__title-link')`**: Mencari elemen `<a>` di dalam artikel yang memiliki kelas tertentu. Elemen ini biasanya merupakan link ke artikel.
- **`href = link.get('href')`**: Mengambil nilai atribut `href` dari elemen `<a>`, yang merupakan URL ke artikel.
- **`title = link.get('title').strip()`**: Mengambil nilai atribut `title` dan menghapus spasi tambahan di awal dan akhir.

### 7. **Memeriksa dan Mengambil Tanggal, Kategori, dan Ringkasan**

- **Memeriksa `href`**: Memastikan bahwa `href` tidak kosong dan belum ada di `seen_links`.
- **Mengambil Tanggal**: Mengambil elemen `<time>` untuk mendapatkan tanggal publikasi, jika ada. Jika tidak ada, tetapkan sebagai 'N/A'.
- **Mengambil Kategori**: Mengambil elemen `<a>` untuk kategori artikel dan menetapkan 'N/A' jika tidak ada.
- **Mengambil Ringkasan**: Mengambil div untuk ringkasan artikel dan menetapkan 'N/A' jika tidak ada.

### 8. **Menyimpan Data ke dalam List**

- Menyimpan data artikel (judul, tanggal, ringkasan, dan kategori) ke dalam list `data_artikel` sebagai satu baris.

### 9. **Menghindari Duplikasi**

- Menambahkan `href` ke set `seen_links` untuk menghindari pengolahan yang sama di masa depan.

### 10. **Membuat DataFrame dari List**

- Membuat DataFrame dari list `data_artikel` dengan kolom yang ditentukan.
